```python
import tarfile
import os
from pathlib import Path

def compress_folder_to_tar_gz(folder_path, output_filename):
    """
    Compress a folder to a .tar.gz file
    
    Args:
        folder_path (str): Path to the folder to compress
        output_filename (str): Name of the output .tar.gz file
    """
    try:
        # Convert to Path objects for easier handling
        folder_path = Path(folder_path)
        output_path = Path(output_filename)
        
        # Check if the folder exists
        if not folder_path.exists():
            print(f"Error: Folder '{folder_path}' does not exist!")
            return False
        
        # Create the tar.gz file
        with tarfile.open(output_path, 'w:gz') as tar:
            # Add the entire folder to the archive
            tar.add(folder_path, arcname=folder_path.name)
        
        print(f"Successfully compressed '{folder_path}' to '{output_path}'")
        
        # Show file size information
        original_size = sum(f.stat().st_size for f in folder_path.rglob('*') if f.is_file())
        compressed_size = output_path.stat().st_size
        compression_ratio = (1 - compressed_size / original_size) * 100 if original_size > 0 else 0
        
        print(f"Original size: {original_size:,} bytes")
        print(f"Compressed size: {compressed_size:,} bytes")
        print(f"Compression ratio: {compression_ratio:.1f}%")
        
        return True
        
    except Exception as e:
        print(f"Error compressing folder: {e}")
        return False

# Compress the 7.6.0B folder
folder_to_compress = "7.6.0B"
output_file = "7.6.0B.tar.gz"

print("Starting compression...")
success = compress_folder_to_tar_gz(folder_to_compress, output_file)

if success:
    print("Compression completed successfully!")
else:
    print("Compression failed!")
```

```python
# Additional utility functions

def list_folder_contents(folder_path, max_files=20):
    """List contents of a folder before compression"""
    folder = Path(folder_path)
    if not folder.exists():
        print(f"Folder '{folder_path}' does not exist!")
        return
    
    print(f"\nContents of '{folder_path}':")
    print("-" * 40)
    
    files = list(folder.rglob('*'))
    total_files = len([f for f in files if f.is_file()])
    total_dirs = len([f for f in files if f.is_dir()])
    
    print(f"Total files: {total_files}")
    print(f"Total directories: {total_dirs}")
    
    if max_files > 0:
        print(f"\nFirst {min(max_files, len(files))} items:")
        for i, item in enumerate(files[:max_files]):
            relative_path = item.relative_to(folder)
            item_type = "DIR" if item.is_dir() else "FILE"
            size = f"({item.stat().st_size:,} bytes)" if item.is_file() else ""
            print(f"  {item_type}: {relative_path} {size}")
        
        if len(files) > max_files:
            print(f"  ... and {len(files) - max_files} more items")

def check_archive_contents(archive_path, max_files=10):
    """Check contents of a tar.gz archive"""
    try:
        with tarfile.open(archive_path, 'r:gz') as tar:
            members = tar.getmembers()
            print(f"\nArchive '{archive_path}' contains {len(members)} items:")
            print("-" * 40)
            
            for i, member in enumerate(members[:max_files]):
                item_type = "DIR" if member.isdir() else "FILE"
                size = f"({member.size:,} bytes)" if member.isfile() else ""
                print(f"  {item_type}: {member.name} {size}")
            
            if len(members) > max_files:
                print(f"  ... and {len(members) - max_files} more items")
                
    except Exception as e:
        print(f"Error reading archive: {e}")

# Check the folder contents before compression (optional)
print("Checking folder contents before compression...")
list_folder_contents("7.6.0B", max_files=10)

# After compression, you can check the archive contents
# Uncomment the line below after running the first cell
# check_archive_contents("7.6.0B.tar.gz", max_files=10)
```
